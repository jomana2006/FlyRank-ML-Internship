```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
DATA = pd.read_csv("/content/content_refresh_anonymized.csv")
DATA.head(20)
```

```python
DATA.columns
```

```python
DATA['ai_sessions_90d'].tolist()
```

```python
DATA['ai_traffic_pct'].to_list()
```

```python
cols_of_interest = ['ai_traffic_pct', 'ai_sessions_90d', 'word_count', 'char_count',
                     'avg_position', 'ctr', 'engagement_rate', 'scroll_rate',
                     'content_age_days', 'days_since_last_update', 'search_volume',
                     'competition', 'cpc', 'sessions_90d', 'impressions_90d']

plt.figure(figsize=(10, 8))
sns.heatmap(DATA[cols_of_interest].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0,0].scatter(DATA['ai_sessions_90d'], DATA['ai_traffic_pct'], alpha=0.5)
axes[0,0].set_xlabel('ai_sessions_90d'); axes[0,0].set_ylabel('ai_traffic_pct')

axes[0,1].scatter(DATA['word_count'], DATA['ai_sessions_90d'], alpha=0.5)
axes[0,1].set_xlabel('word_count'); axes[0,1].set_ylabel('ai_sessions_90d')

axes[1,0].scatter(DATA['sessions_90d'], DATA['ai_sessions_90d'], alpha=0.5)
axes[1,0].set_xlabel('sessions_90d'); axes[1,0].set_ylabel('ai_sessions_90d')

axes[1,1].scatter(DATA['impressions_90d'], DATA['ai_traffic_pct'], alpha=0.5)
axes[1,1].set_xlabel('impressions_90d'); axes[1,1].set_ylabel('ai_traffic_pct')

plt.tight_layout()
plt.show()
```
